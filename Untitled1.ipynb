```python
import re
import time
import os
import requests
import pandas as pd
from bs4 import BeautifulSoup
from datetime import date, timedelta


cinemas = {
    "Ангара", "Берёзка", "Вымпел", "Жуковский", "Искра",
    "Кинопарк", "Космос", "Марс", "Молодёжный", "Нева",
    "Рассвет", "Салют", "Сатурн", "Спутник", "Тула",
    "Факел", "Эльбрус", "Юность"
}


def get_lines(day):
    url = f"https://mos-kino.ru/schedule/?date={day}"

    r = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=20
    )
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    text = soup.get_text("\n")

    lines = [re.sub(r"\s+", " ", x).strip() for x in text.split("\n")]
    lines = [x for x in lines if x]

    return lines, url


def is_info_line(x):
    return bool(re.match(r"^\d{4}\s*/\s*.*?\s*мин\s*/\s*.*?\s*/\s*\d+\+$", x))


def is_time(x):
    return bool(re.match(r"^[0-2]\d:[0-5]\d$", x))


def is_price(x):
    return bool(re.match(r"^\d+\s*[PР₽]$", x))


def is_address(x):
    x_low = x.lower()

    address_patterns = [
        r"\bул\.",
        r"\bулица\b",
        r"\bб-р\b",
        r"\bбульвар\b",
        r"\bпроспект\b",
        r"\bпр-т\b",
        r"\bшоссе\b",
        r"\bнабережная\b",
        r"\bпереулок\b",
        r"\bпер\.",
        r"\bплощадь\b",
        r"\bд\.\s*\d+",
        r"\bдом\s+\d+",
        r"\bцентр\s+«",
        r"\bрайонный центр\b",
    ]

    return any(re.search(p, x_low) for p in address_patterns)


def parse_info(x):
    m = re.match(
        r"^(\d{4})\s*/\s*(.*?)\s*мин\s*/\s*(.*?)\s*/\s*(\d+\+)$",
        x
    )

    if not m:
        return None, None, None, None

    year = int(m.group(1))

    duration_raw = m.group(2).strip()
    duration = int(duration_raw) if duration_raw.isdigit() else None

    country = m.group(3).strip() or None
    age = m.group(4).strip() or None

    return year, duration, country, age


def parse_day(day):
    lines, source_url = get_lines(day)

    rows = []

    cinema = None
    address = None
    metro = []
    movie_title = None
    movie_info = None

    i = 0

    while i < len(lines):
        x = lines[i]

        if x in cinemas:
            cinema = x
            address = None
            metro = []
            movie_title = None
            movie_info = None
            i += 1
            continue

        if is_address(x):
            address = x
            i += 1
            continue

        if x == "●":
            if i + 1 < len(lines):
                metro.append(lines[i + 1])
                i += 2
                continue

        if i + 1 < len(lines) and is_info_line(lines[i + 1]):
            movie_title = x
            movie_info = parse_info(lines[i + 1])
            i += 2
            continue

        if is_time(x) and movie_title and movie_info:
            session_time = x
            fmt = None
            has_subtitles = False
            price = None

            j = i + 1

            while j < len(lines):
                y = lines[j]

                if y in cinemas:
                    break

                if is_info_line(y):
                    break

                if is_time(y):
                    break

                if j + 1 < len(lines) and is_info_line(lines[j + 1]):
                    break

                if y in ["2D", "3D"]:
                    fmt = y

                if y == "СУБ":
                    has_subtitles = True

                if is_price(y):
                    price = int(re.findall(r"\d+", y)[0])
                    break

                j += 1

            year, duration, country, age = movie_info

            rows.append({
                "date": str(day),
                "cinema": cinema,
                "address": address,
                "metro": ", ".join(metro) if metro else None,
                "movie_title": movie_title,
                "year": year,
                "duration_min": duration,
                "country": country,
                "age_rating": age,
                "session_time": session_time,
                "format": fmt,
                "has_subtitles": has_subtitles,
                "price": price,
                "source_url": source_url
            })

            i = j + 1
            continue

        i += 1

    return pd.DataFrame(rows)


def parse_period(start_day, days=7):
    dfs = []

    for k in range(days):
        day = start_day + timedelta(days=k)
        print("Парсим:", day)

        try:
            df_day = parse_day(day)
            print("  найдено сеансов:", len(df_day))
            dfs.append(df_day)
        except Exception as e:
            print("  ошибка:", e)

        time.sleep(0.5)

    if dfs:
        return pd.concat(dfs, ignore_index=True)

    return pd.DataFrame()


start_day = date(2026, 5, 18)
df_all = parse_period(start_day, days=7)

df_foreign = df_all[
    ~df_all["country"].str.lower().str.contains("россия", na=False)
].copy()

print("Все сеансы:", df_all.shape)
print("Зарубежные сеансы:", df_foreign.shape)

print("\nСеансы по датам:")
print(df_foreign.groupby("date").size())

print("\nТоп фильмов:")
print(df_foreign["movie_title"].value_counts().head(20))

os.makedirs("data/raw", exist_ok=True)
os.makedirs("data/processed", exist_ok=True)

df_all.to_csv("data/raw/moskino_schedule_week_all.csv", index=False, encoding="utf-8-sig")
df_foreign.to_csv("data/processed/moskino_schedule_week_foreign.csv", index=False, encoding="utf-8-sig")

df_foreign.head(30)
```

```python
import pandas as pd
import os

file_path = "/content/data-14-structure-5.csv"

df_reg = pd.read_csv(
    file_path,
    encoding="utf-8",
    sep=",",
    low_memory=False
)

df_reg.columns = [c.strip() for c in df_reg.columns]

cols = {
    "Название фильма": "movie_title",
    "Hаименование на иностранном языке": "foreign_title",
    "Номер удостоверения": "certificate_number",
    "Дата регистрации удостоверения": "certificate_registration_date",
    "Дата начала показа фильма": "release_start_date",
    "Страна производства": "country",
    "Год производства": "year",
    "Режиссер": "director",
    "Студия-производитель": "studio",
    "Права на использование фильма": "rights_holder",
    "Срок действия прав проката до": "rights_valid_until",
    "Возрастная категория": "age_category",
    "Запись удалена": "is_deleted"
}

existing_cols = [c for c in cols.keys() if c in df_reg.columns]

df_clean = df_reg[existing_cols].rename(columns=cols).copy()

if "movie_title" in df_clean.columns:
    df_clean["movie_title"] = (
        df_clean["movie_title"]
        .astype(str)
        .str.strip()
        .str.lower()
    )
    df_clean.loc[df_clean["movie_title"].isin(["nan", "none", ""]), "movie_title"] = None

if "foreign_title" in df_clean.columns:
    df_clean["foreign_title"] = (
        df_clean["foreign_title"]
        .astype(str)
        .str.strip()
        .str.lower()
    )
    df_clean.loc[df_clean["foreign_title"].isin(["nan", "none", ""]), "foreign_title"] = None

if "country" in df_clean.columns:
    df_clean["country"] = df_clean["country"].astype(str).str.strip()
    df_clean.loc[df_clean["country"].isin(["nan", "None", ""]), "country"] = None

if "year" in df_clean.columns:
    df_clean["year"] = (
        df_clean["year"]
        .astype(str)
        .str.extract(r"(\d{4})")[0]
    )
    df_clean["year"] = pd.to_numeric(df_clean["year"], errors="coerce").astype("Int64")

for col in ["certificate_registration_date", "release_start_date", "rights_valid_until"]:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce").dt.date

if "is_deleted" in df_clean.columns:
    df_clean = df_clean[
        df_clean["is_deleted"].isna()
        | (df_clean["is_deleted"].astype(str).str.strip() == "")
    ].copy()

df_clean = df_clean.drop_duplicates()

print(df_clean.shape)
df_clean.head()
```
